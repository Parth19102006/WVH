```python
import pandas as pd
import pickle

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report


# =========================================================
# 1. LOAD DATA
# =========================================================

df = pd.read_csv("synthetic_login_data_final_audited.csv")


# =========================================================
# 2. FEATURES
# =========================================================

numeric_features = [
    "login_hour",
    "is_new_device",
    "is_new_country",
    "location_change",
    "ip_risk_score",
    "failed_login_attempts",
    "time_since_last_login_hrs",
    "login_frequency_per_day",
    "historical_behavior_score",
    "is_first_login"
]

categorical_features = [
    "device",
    "device_type",
    "browser",
    "country"
]

features = numeric_features + categorical_features

X = df[features]

# Used ONLY for stratification/evaluation
y = df["risk_level"]


# =========================================================
# 3. PREPROCESSING
# =========================================================
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])



# =========================================================
# 4. ISOLATION FOREST
# =========================================================

isolation_forest = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)


# =========================================================
# 5. COMPLETE PIPELINE
# =========================================================

pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", isolation_forest)
])


# =========================================================
# 6. STRATIFIED K-FOLD
# =========================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold_scores = []

for fold, (train_index, test_index) in enumerate(skf.split(X, y), 1):

    X_train = X.iloc[train_index]
    X_test = X.iloc[test_index]

    y_test = y.iloc[test_index]

    # Train
    pipeline.fit(X_train)

    # Predict
    predictions = pipeline.predict(X_test)

    # Isolation Forest:
    #  1  = normal
    # -1  = anomaly

    predictions = [1 if p == -1 else 0 for p in predictions]

    # Treat High/Critical as suspicious
    actual = y_test.isin(["High", "Critical"]).astype(int)

    # Calculate how many suspicious logins were detected
    detected = sum(
        (predictions[i] == 1 and actual.iloc[i] == 1)
        for i in range(len(predictions))
    )

    total_suspicious = actual.sum()

    recall = detected / total_suspicious if total_suspicious > 0 else 0

    fold_scores.append(recall)

    print(f"Fold {fold} Recall: {recall:.3f}")


# =========================================================
# 7. CROSS-VALIDATION RESULT
# =========================================================

print("\nCross-validation results:")
print("Average Recall:", sum(fold_scores) / len(fold_scores))


# =========================================================
# 8. TRAIN FINAL MODEL ON ALL DATA
# =========================================================

pipeline.fit(X)


# =========================================================
# 9. SAVE MODEL
# =========================================================

with open("isolation_forest_pipeline.pkl", "wb") as file:
    pickle.dump(pipeline, file)


# =========================================================
# 10. SAVE FEATURE LIST
# =========================================================

with open("features.pkl", "wb") as file:
    pickle.dump(features, file)


print("\nModel saved as isolation_forest_pipeline.pkl")
print("Features saved as features.pkl")
```
